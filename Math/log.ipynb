```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import random
import math
max_ind = 1000
x_list = []
y_list = []
for _ in range(10):
    num = random.randint(0,max_ind)
    x_list.append([float(num)])
    log_num = math.log(num)
    print(log_num)
    y_list.append(log_num)
x_norm = max(x_list)
y_norm = max(y_list)
```

```python
class calc(nn.Module):
    def __init__(self):
        super().__init__()
        self.mlp = nn.Linear(1,1024)
        self.act = nn.ReLU()
        self.out = nn.Linear(1024,1)
    def forward(self,x):
        x = self.mlp(x)
        x = self.act(x)
        x = self.out(x)
        return x
    
model = calc()
optimizer = torch.optim.AdamW(model.parameters(),lr=1e-4)
loss_type = nn.MSELoss()
```

```python
for _ in range(10000):
    index = random.randint(0,len(x_list))
    x = torch.tensor(x_list[index],dtype=torch.float)
    y = torch.tensor(y_list[index],dtype=torch.float)
    logits = model(x)
    
    loss = loss_type(logits,y) 

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    if _ % 100 == 0:
        print(f"step: {_} loss: {loss.item()}")
```

```python
def generate(start):
    logits = model(start)
    print(logits*max_ind)
generate(torch.tensor([1/max_ind,100/max_ind],dtype=torch.float))



```
