```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import random
max_ind = 90000
x_list = []
y_list = []
for _ in range(80000):
    num1 = random.randint(0,max_ind)
    num2 = random.randint(0,max_ind)
    x_list.append([num1/max_ind,num2/max_ind])
    sum = num1 + num2
    y_list.append(sum / (2 * max_ind))
```

```python
class calc(nn.Module):
    def __init__(self):
        super().__init__()
        self.out = nn.Linear(2, 1, bias=False)
    def forward(self, x):
        return self.out(x)
    
model = calc()
optimizer = torch.optim.AdamW(model.parameters(),lr=1e-4)
loss_type = nn.MSELoss()
```

```python
def train(index,x_train,y_train):
    model.train()
    x = torch.tensor(x_train[index],dtype=torch.float)
    y = torch.tensor(y_train[index],dtype=torch.float)
    logits = model(x)
    
    loss = loss_type(logits,y) 

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    if _ % 100 == 0:
        print(f"step: {_} loss: {loss.item()*max_ind}")
for _ in range(40000):
    index = random.randint(0, len(x_list) - 1)
    train(index,x_list,y_list)
```

```python
correction_steps = 100
def generate(start):
    model.eval()
    y = start[0]*max_ind + start[1]*max_ind
    logits = model(start)
    logits = logits * ( 2 * max_ind)
    logits = logits.tolist()
    error = int(logits[0])-int(y)
     
    print(f"Answer: {int(logits[0])} Error: {error} Correct Answer: {int(logits[0])-error}")
generate(torch.tensor([100/max_ind,104/max_ind],dtype=torch.float))



```
