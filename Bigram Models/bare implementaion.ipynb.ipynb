```python
from torch._C import dtype
import torch
import torch.nn as nn
import torch.nn.functional as F

with open("/content/sample_data/README.md","r") as f:
  text = f.read()
chars = sorted(set(text))
stoi = {s:i for i,s in enumerate(chars)}
itos = {i:s for i,s in enumerate(chars)}
encode = lambda s: [stoi[c] for c in s]
decode = lambda l: ''.join([itos[i] for i in l])

text = encode(text)
x = []
y = []

for i in range(500):
  x_c = text[i]
  y_c = text[i+1]

  x_c = torch.tensor(x_c,dtype=torch.long)
  y_c = torch.tensor(y_c,dtype=torch.long)

  x_c = F.one_hot(x_c,num_classes=len(chars)).float()
  y_c = F.one_hot(y_c,num_classes=len(chars)).float()
  x.append(x_c)
  y.append(y_c)
x = torch.stack(x)
y = torch.stack(y)
```

```python
class NuralNetwork(nn.Module):
  def __init__(self,inp_size,H_size,out_size):
    super().__init__()
    self.emb = nn.Parameter(torch.randn(inp_size, 512)* 0.01)
    self.w1 = nn.Parameter(torch.randn(512, H_size) * 0.01)
    self.b1 = nn.Parameter(torch.zeros(H_size))
    self.w2 = nn.Parameter(torch.randn(H_size, out_size) * 0.01)
    self.b2 = nn.Parameter(torch.zeros(out_size))
  def forward(self,x):

    out = x @ self.emb
    out = out @ self.w1 + self.b1
    out = torch.relu(out)
    out = out @ self.w2 + self.b2
    out = torch.relu(out)
    return out


```

```python
AI = NuralNetwork(len(chars),512,len(chars))
for _ in range(6000):
  logits = AI(x)
  loss = F.cross_entropy(logits,y)
  loss.backward()
  for p in AI.parameters():
    p.data += -0.1 * p.grad
  AI.zero_grad()
  print(f"\rstep: {_} | loss: {loss.item()} ",end="",flush=True)
```

```python
def generate(x,length):
  generation = x
  x = F.one_hot(x,num_classes=len(chars)).float()
  for l in range(length):
    logits = AI(x)
    logits = logits
    x = F.softmax(logits,dim=-1)
    logits = torch.multinomial(x[0],num_samples=1,replacement=True)
    generation = torch.cat([generation, logits], dim=0)
  print(decode(generation.tolist()))
x = torch.tensor(encode("hello"),dtype=torch.long)

generate(x,100)
```
